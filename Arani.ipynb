```python
#What are the daily, weekly, and monthly sales patterns?
# Convert order_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'])
# Daily Sales Trend
daily_sales = df.groupby(df['order_date'].dt.date)['total_price'].sum()

# Weekly Sales Trend
weekly_sales = df.groupby(df['order_date'].dt.isocalendar().week)['total_price'].sum()

# Monthly Sales Trend
monthly_sales = df.groupby(df['order_date'].dt.month)['total_price'].sum()
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
daily_sales.plot(title="Daily Sales Trend")
plt.show()

weekly_sales.plot(title="Weekly Sales Trend")
plt.show()

monthly_sales.plot(title="Monthly Sales Trend")
plt.show()
```

```python
#Which pizza types generate the highest revenue and volume?
pizza_revenue = df.groupby('pizza_name')['total_price'].sum()
pizza_volume = df.groupby('pizza_name')['quantity'].sum()
print(pizza_revenue)
print(pizza_volume)
```

```python
#Do customers prefer certain sizes or toppings more often?
pizza_size_sales = df.groupby('pizza_size')['quantity'].sum()
pizza_topping_sales = df.groupby('pizza_ingredients')['quantity'].sum()
print(pizza_size_sales)
print(pizza_topping_sales)
```

```python
#At what times of day are sales highest?
df['order_hour'] = df['order_time'].apply(lambda x: x.hour)
pizza_hour_sales = df.groupby('order_hour')['quantity'].sum()
print(pizza_hour_sales)
```

```python
#Which factors (size, category, location) contribute most to revenue growth?
pizza_size = df.groupby('pizza_size')['total_price'].sum()
pizza_category = df.groupby('pizza_category')["total_price"].sum()

print(pizza_size)
print(pizza_category)
```

```python
# Which pizzas have the best profit margins compared to their sales volume?
pizza_profit = df.groupby('pizza_name')['total_price'].sum().idxmax()

pizza_volume = df.groupby('pizza_name')['quantity'].sum().idxmax()
print(pizza_profit)
print(pizza_volume)
```

```python
#Revenue by Category
df['revenue'] = df['quantity']* df['unit_price']
revenue_by_category = df.groupby('pizza_category')['revenue'].sum()
print(df['revenue'])
print(revenue_by_category)
```

```python
#Peak Sales Hour
peak_sales_hour = df.groupby('order_hour')['quantity'].sum().idxmax()
print(peak_sales_hour)
```

```python
#Profit Margin
# NOTE: df['cost_price'] is currently calculated as the sum of all unit prices, which is likely incorrect
# for individual item profit margin. A proper 'cost' column would be needed for accurate profit.
df['cost_price'] = df['unit_price'].sum()
print(df['cost_price'])
df['profit_margin'] = df['revenue'] - df['cost_price'] * df['revenue'] * 100
print(df['profit_margin'])
```

```python
# Assuming a standard food cost of 30% (70% profit margin) since actual cost data is not available
df['profit_margin'] = df['total_price'] * 0.70

Total_profit = df['profit_margin'].sum()
print(f"Estimated Total Profit (at 70% margin): ${Total_profit:,.2f}")
```

```python
# Recalculate revenue by category to ensure it exists
revenue_by_category = df.groupby('pizza_category')['total_price'].sum()

# Get total revenue
total_revenue = df['total_price'].sum()

# Calculate and print category share
category_share = (revenue_by_category / total_revenue) * 100
print(category_share)
```
